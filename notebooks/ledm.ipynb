```python
import numpy as np
import matplotlib.pyplot as plt
from classy import Class

def compute_pk(params, z=0.0, kmin=1e-4, kmax=1, nk=200):
    cosmo = Class()
    cosmo.set(params)
    cosmo.compute()

    h = cosmo.h()
    k = np.logspace(np.log10(kmin), np.log10(kmax), nk)
    pk = np.array([cosmo.pk(kk*h, z) * h**3 for kk in k])  # Convert units

    cosmo.struct_cleanup()
    cosmo.empty()
    return k, pk
```

```python
# ΛCDM baseline params
params_LCDM = {
    'output': 'mPk',
    'P_k_max_1/Mpc': 10,
    'z_max_pk': 5.0,
    'h': 0.67,
    'omega_b': 0.02236,
    'omega_cdm': 0.12,
    'A_s': 2.101e-9,
    'n_s': 0.9649,
    'tau_reio': 0.0544
}

# eDM parameters (best-fit)
params_eDM = params_LCDM.copy()
params_eDM.update({
    'w_0_eDM': -0.3,
    'a_nz_eDM': 0.5,
    'output': 'mPk'
})
```

```python
# ΛCDM baseline params
params_LCDM = {
    'output': 'mPk',
    'P_k_max_1/Mpc': 10,
    'z_max_pk': 5.0,
    'h': 0.67,
    'omega_b': 0.02236,
    'omega_cdm': 0.12,
    'A_s': 2.101e-9,
    'n_s': 0.9649,
    'tau_reio': 0.0544
}

# eDM parameters (best-fit)
params_eDM = params_LCDM.copy()
params_eDM.update({
    'w_0_eDM': -0.3,
    'a_nz_eDM': 0.5,
    'output': 'mPk'
})
```

```python
k_LCDM, Pk_LCDM = compute_pk(params_LCDM)
k_eDM, Pk_eDM = compute_pk(params_eDM)

plt.figure(figsize=(8,6))
plt.loglog(k_LCDM, Pk_LCDM, label='ΛCDM', c='k', lw=2)
plt.loglog(k_eDM, Pk_eDM, label='eDM ($w_0=-0.3$, $a_{nz}=0.5$)', c='tab:red', lw=2)
plt.xlabel(r'$k$ [h Mpc$^{-1}$]')
plt.ylabel(r'$P(k)$ [(Mpc/h)$^3$]')
plt.title('Linear Matter Power Spectrum Comparison')
plt.legend()
plt.grid(True, which='both', ls='--', alpha=0.5)
plt.show()
```
